```python
import streamlit as st
import pickle
import numpy as np

# Load your trained model (ensure 'model.pkl' is in the same folder)
with open('best_model.pkl', 'rb') as f:
    model = pickle.load(f)

st.title("Autism Disorder Prediction")
st.markdown("Enter your details to get a prediction:")

# Example input fields - adjust these as needed
age = st.number_input("Age", min_value=1, max_value=120)
gender = st.selectbox("Gender", ["Male", "Female"])
a1 = st.number_input("A1 Screening Score", min_value=0, max_value=10)
a2 = st.number_input("A2 Screening Score", min_value=0, max_value=10)
a3 = st.number_input("A3 Screening Score", min_value=0, max_value=10)
a4 = st.number_input("A4 Screening Score", min_value=0, max_value=10)
a5 = st.number_input("A5 Screening Score", min_value=0, max_value=10)

# Prepare input for prediction (ensure this matches your training data shape)
input_data = np.array([[age, 1 if gender=="Male" else 0, a1, a2, a3, a4, a5]])

if st.button("Predict"):
    prediction = model.predict(input_data)
    st.success(f"Prediction: {'Autism detected' if prediction==1 else 'No Autism'}")
```
